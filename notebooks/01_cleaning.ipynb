```python
df = pd.read_csv('C:/Users/LENOVO/layoffs-data-analysis/data/raw/layoffs.csv')
df['date'] = pd.to_datetime(df['date'], format='mixed', dayfirst=False)
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed', dayfirst=False)
df[['date', 'date_added']].isnull().sum()
```

```python
df['date'] = df['date'].dt.strftime('%Y-%m-%d')
df['date_added'] = df['date_added'].dt.strftime('%Y-%m-%d')
df[['date', 'date_added']].head()
```

```python
df.dtypes
```

```python
df['company'] = df['company'].str.strip()
df['industry'].unique()
```

```python
df['country'].unique()
```

```python
df['stage'].unique()
```

```python
df['country'] = df['country'].replace('UAE', 'United Arab Emirates')
df['country'].nunique()
```

```python
df = df.dropna(subset=['location','industry','source','stage','country'])
df.shape
```

```python
df[['total_laid_off', 'funds_raised']].describe()
```

```python
df['percentage_laid_off'].describe()
```

```python
df['company'].nunique(), df['company'].str.lower().nunique()
```

```python
df['company_lower'] = df['company'].str.lower()
dupes = df.groupby('company_lower')['company'].unique()
dupes[dupes.apply(len) > 1]
```

```python
canonical_names = df.groupby('company_lower')['company'].first()
df['company'] = df['company_lower'].map(canonical_names)
df['company'].nunique()
```

```python
df = df.drop(columns=['company_lower'])
```
